```python
from nltk.stem import PorterStemmer

pst = PorterStemmer()
```

```python
pst.stem("affection")
```

```python
pst.stem('playing')
```

```python
pst.stem('fucking')
```

```python
pst.stem('station')
```

```python
words_to_stem = [
    'give', 'giving', 'given', 'gaved',
    'thinking', 'loving', 'maximum', 'Hello'
]

for word in words_to_stem:
    print(word + " : " + pst.stem(word))
```

```python
from nltk.stem import SnowballStemmer

sbst = SnowballStemmer("english")

for word in words_to_stem:
    print(word + " : " + sbst.stem(word))
```
