```python
from sklearn.model_selection import train_test_split
from sklearn.impute import SimpleImputer
from sklearn.compose import ColumnTransformer
```

```python
df = pd.read_csv(r"E:\archive\tested.csv")
```

```python
df
```

```python
df.isnull().mean()
```

```python
x = df.drop(columns=['Survived'])
y = df['Survived']
```

```python
x_train,x_test, y_train, y_test = train_test_split(x,y, test_size=0.2, random_state=2)
```

```python
x_train.shape, x_test.shape
```

```python
x_train.isnull().mean()
```

```python
mean_age = x_train['Age'].mean()
median_age = x_train['Age'].median()

mean_fare = x_train['Fare'].mean()
median_fare = x_train['Fare'].median()
```

```python
x_train['Age_median'] = x_train['Age'].fillna(median_age)
x_train['Age_mean'] = x_train['Age'].fillna(mean_age)
x_train['Fare_mean'] = x_train['Fare'].fillna(mean_age)
```

```python
x_train.sample(5)
```

```python
print("Original Age variable variance:", x_train['Age'].var())

print("Age variance after median imputation:",
      x_train['Age_median'].var())

print("Age variance after mean imputation:",
      x_train['Age_mean'].var())

print("Original Fare variable variance:",
      x_train['Fare'].var())

print("Fare variance after median imputation:",
      x_train['Fare_median'].var())

print("Fare variance after mean imputation:",
      x_train['Fare_mean'].var())
```

```python
fig = plt.figure()
ax = fig.add_subplot(111)

# original variable distribution 
x_train['Age'].plot(kind = 'kde', ax = ax)

# variable imputed with the median
x_train['Age_median'].plot(kind='kde', ax = ax,  color='red')
```

```python
import matplotlib.pyplot as plt

fig = plt.figure()
ax = fig.add_subplot(111)

# Original variable distribution
x_train['Fare_median'].plot(
    kind='kde',
    ax=ax,
    color='red'
)

# Variable imputed with the mean
x_train['Fare_mean'].plot(
    kind='kde',
    ax=ax,
    color='green'
)

# Add legends
lines, labels = ax.get_legend_handles_labels()
ax.legend(lines, labels, loc='best')

plt.show()
```

```python
import numpy as np
import pandas as pd

from sklearn.model_selection import train_test_split

from sklearn.impute import KNNImputer, SimpleImputer

from sklearn.linear_model import LogisticRegression

from sklearn.metrics import accuracy_score
```

```python

df = pd.read_csv(r"E:\archive\tested.csv")[['Age', 'Pclass', 'Fare', 'Survived']]
```

```python
df
```

```python
df.isnull().mean()*100
```

```python
x = df.drop(columns=['Survived'])
y =  df['Survived']
```

```python
x_train,x_test, y_train,y_test = train_test_split(x,y, test_size=0.2, random_state=2)
```

```python
x_train.head()
```

```python
knn = KNNImputer()

x_train_trf = knn.fit_transform(x_train)
x_test_trf = knn.transform(x_test)
```

```python
pd.DataFrame(x_train_trf, columns=x_train.columns)
```

```python
from sklearn.impute import KNNImputer

knn = KNNImputer(n_neighbors=5, weights='distance')

x_train_trf = knn.fit_transform(x_train)
x_test_trf = knn.transform(x_test)
```

```python
lr = LogisticRegression()
lr.fit(x_train_trf, y_train)
y_pred = lr.predict(x_test_trf)
accuracy_score(y_test,y_pred)
```

```python
# comparision with simple Imputer --> mean
si = SimpleImputer()

x_train_trf2 = si.fit_transform(x_train)
x_test_trf2 = si.transform(x_test)
```

```python
lr =LogisticRegression()
lr.fit_trf2 = si.fit_transform(x_train)
x_test_trf2 = si.transform(x_test)
```

```python
lr = LogisticRegression()
lr.fit(x_train_trf2,y_train)
y_pred2 = lr.predict(x_test_trf2)
accuracy_score(y_test,y_pred2)
```
