# Capping

```python
upper_limit = df['CGPA'].mean() + 3*df['CGPA'].std()
lower_limit = df['CGPA'].mean() -3*df['CGPA'].std()
```

```python
upper_limit
```

```python
lower_limit
```

```python
df['CGPA'] = np.where(
    df['CGPA']> upper_limit,
    upper_limit,
    np.where(
        df['CGPA']<lower_limit,
        lower_limit,
        df['CGPA']
    )
)    
```
