```python
import os
from tensorflow.keras.preprocessing.image import ImageDataGenerator, load_img, img_to_array

# Create ImageDataGenerator
datagen = ImageDataGenerator(
    rotation_range=40,
    width_shift_range=0.2,
    height_shift_range=0.2,
    shear_range=0.2,
    zoom_range=0.2,
    horizontal_flip=True,
    fill_mode="nearest"
)

# Input folder containing airplane images
input_dir = r"E:\archive (1)\data\natural_images\airplane"

# Output folder for augmented images
output_dir = r"E:\archive (1)\data\natural_images\airplane_augmented"

os.makedirs(output_dir, exist_ok=True)

# Get all image files
image_files = [
    f for f in os.listdir(input_dir)
    if f.lower().endswith((".jpg", ".jpeg", ".png"))
]

print("Number of original images:", len(image_files))

# Process each image
for filename in image_files:

    # Load image
    img_path = os.path.join(input_dir, filename)
    img = load_img(img_path)

    # Convert image to NumPy array
    x = img_to_array(img)

    # Reshape
    x = x.reshape((1,) + x.shape)

    # Generate 5 augmented images for each original image
    i = 0

    for batch in datagen.flow(
        x,
        batch_size=1,
        save_to_dir=output_dir,
        save_prefix="aug_" + os.path.splitext(filename)[0],
        save_format="jpeg"
    ):
        i += 1

        if i >= 5:
            break

print("Augmentation completed!")
```

```python
import os
import glob
import numpy as np

from tensorflow.keras.applications.resnet50 import (
    ResNet50,
    preprocess_input,
    decode_predictions
)

from tensorflow.keras.preprocessing import image


# --------------------------------
# 1. Find airplane images
# --------------------------------

airplane_folder = r"E:\archive (1)\data\natural_images\airplane"

images = glob.glob(airplane_folder + r"\*.jpg")

print("Number of images:", len(images))

if len(images) == 0:
    print("No JPG images found!")
else:

    # Select first image
    img_path = images[0]

    print("Image selected:")
    print(img_path)


    # --------------------------------
    # 2. Load ResNet50
    # --------------------------------

    model = ResNet50(weights="imagenet")


    # --------------------------------
    # 3. Load image
    # --------------------------------

    img = image.load_img(
        img_path,
        target_size=(224, 224)
    )


    # --------------------------------
    # 4. Convert image to NumPy array
    # --------------------------------

    img_array = image.img_to_array(img)

    print("Image shape:", img_array.shape)


    # --------------------------------
    # 5. Add batch dimension
    # --------------------------------

    img_array = np.expand_dims(
        img_array,
        axis=0
    )

    print("Batch shape:", img_array.shape)


    # --------------------------------
    # 6. Preprocess
    # --------------------------------

    img_array = preprocess_input(img_array)


    # --------------------------------
    # 7. Prediction
    # --------------------------------

    predictions = model.predict(img_array)


    # --------------------------------
    # 8. Decode prediction
    # --------------------------------

    decoded_predictions = decode_predictions(
        predictions,
        top=5
    )[0]


    print("\nPredictions:")

    for i, (imagenet_id, label, score) in enumerate(
        decoded_predictions
    ):
        print(
            f"{i + 1}: {label} ({score:.2f})"
        )
```

```python
import os
import glob
import numpy as np

from tensorflow.keras.applications.resnet50 import (
    ResNet50,
    preprocess_input,
    decode_predictions
)

from tensorflow.keras.preprocessing import image


# Load ResNet50 model pre-trained on ImageNet
model = ResNet50(weights="imagenet")


# ------------------------------------------------
# Find airplane images
# ------------------------------------------------

airplane_folder = r"E:\archive (1)\data\natural_images\airplane"

image_files = []

for extension in ["*.jpg", "*.jpeg", "*.png", "*.JPG", "*.JPEG", "*.PNG"]:
    image_files.extend(
        glob.glob(os.path.join(airplane_folder, extension))
    )


# Check images
print("Number of airplane images:", len(image_files))

if len(image_files) == 0:
    print("No airplane images found!")
else:

    # Select the first airplane image
    img_path = image_files[0]

    print("Selected image:")
    print(img_path)


    # ------------------------------------------------
    # Load image
    # ------------------------------------------------

    img = image.load_img(
        img_path,
        target_size=(224, 224)
    )


    # ------------------------------------------------
    # Convert image to NumPy array
    # ------------------------------------------------

    img_array = image.img_to_array(img)

    print("Image shape:", img_array.shape)


    # ------------------------------------------------
    # Add batch dimension
    # ------------------------------------------------

    img_array = np.expand_dims(
        img_array,
        axis=0
    )

    print("Batch shape:", img_array.shape)


    # ------------------------------------------------
    # Preprocess image
    # ------------------------------------------------

    img_array = preprocess_input(img_array)


    # ------------------------------------------------
    # Prediction
    # ------------------------------------------------

    predictions = model.predict(img_array)


    # ------------------------------------------------
    # Decode prediction
    # ------------------------------------------------

    decoded_predictions = decode_predictions(
        predictions,
        top=5
    )[0]


    print("\nPredictions:")

    for i, (imagenet_id, label, score) in enumerate(
        decoded_predictions
    ):
        print(
            f"{i + 1}: {label} ({score:.2f})"
        )


    # Top prediction
    top_class_index = np.argmax(predictions[0])

    print("\nTop Prediction Class Index:", top_class_index)
```

```python
import os
import glob
import numpy as np

from tensorflow.keras.applications.resnet_v2 import (
    ResNet50V2,
    preprocess_input,
    decode_predictions
)

from tensorflow.keras.preprocessing import image


# ==========================================
# 1. Load ResNet50V2 model
# ==========================================

model = ResNet50V2(weights="imagenet")


# ==========================================
# 2. Airplane folder
# ==========================================

airplane_folder = r"E:\archive (1)\data\natural_images\airplane"


# ==========================================
# 3. Find an actual image
# ==========================================

image_files = []

for extension in [
    "*.jpg",
    "*.jpeg",
    "*.png",
    "*.JPG",
    "*.JPEG",
    "*.PNG"
]:
    image_files.extend(
        glob.glob(
            os.path.join(
                airplane_folder,
                extension
            )
        )
    )


print("Number of images:", len(image_files))


# ==========================================
# 4. Check if image exists
# ==========================================

if len(image_files) == 0:

    print("No airplane image found!")

else:

    # Select first airplane image
    img_path = image_files[0]

    print("Selected image:")
    print(img_path)


    # ==========================================
    # 5. Load image
    # ==========================================

    img = image.load_img(
        img_path,
        target_size=(224, 224)
    )


    # ==========================================
    # 6. Convert image to NumPy array
    # ==========================================

    img_array = image.img_to_array(img)

    print("Image shape:", img_array.shape)


    # ==========================================
    # 7. Add batch dimension
    # ==========================================

    img_array = np.expand_dims(
        img_array,
        axis=0
    )


    # ==========================================
    # 8. Preprocess image
    # ==========================================

    img_array = preprocess_input(
        img_array
    )


    # ==========================================
    # 9. Make prediction
    # ==========================================

    predictions = model.predict(
        img_array
    )


    # ==========================================
    # 10. Decode predictions
    # ==========================================

    decoded_predictions = decode_predictions(
        predictions,
        top=5
    )[0]


    print("\nPredictions:")

    for i, (imagenet_id, label, score) in enumerate(
        decoded_predictions
    ):
        print(
            f"{i + 1}: {label} ({score:.2f})"
        )


    # ==========================================
    # 11. Top prediction class index
    # ==========================================

    top_class_index = np.argmax(
        predictions[0]
    )

    print(
        "\nTop Prediction Class Index:",
        top_class_index
    )
```

# VGG16


```python
import numpy as np

from tensorflow.keras.applications.vgg16 import (
    VGG16,
    preprocess_input,
    decode_predictions
)

from tensorflow.keras.preprocessing import image


# Load VGG16 model pre-trained on ImageNet
model = VGG16(weights="imagenet")


# Image path
img_path = r"C:\Users\srava\Downloads\mahindra-thar-2020-black.jpg"


# Load image
img = image.load_img(
    img_path,
    target_size=(224, 224)
)


# Convert image to NumPy array
img_array = image.img_to_array(img)


# Add batch dimension
img_array = np.expand_dims(
    img_array,
    axis=0
)


# Preprocess image for VGG16
img_array = preprocess_input(
    img_array
)


# Make predictions
predictions = model.predict(
    img_array
)


# Decode top-3 predictions
decoded_predictions = decode_predictions(
    predictions,
    top=3
)[0]


# Print predictions
print("Predictions:")

for i, (imagenet_id, label, score) in enumerate(
    decoded_predictions
):
    print(
        f"{i + 1}: {label} ({score:.2f})"
    )


# Get top prediction class index
top_class_index = np.argmax(
    predictions[0]
)

print(
    f"\nTop Prediction Class Index: {top_class_index}"
)
```
