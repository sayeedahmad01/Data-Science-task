```python
import matplotlib.pyplot as plt

x = x[:, 0]

plt.scatter(x, y)
plt.xlabel("Feature X")
plt.ylabel("Target y")
plt.title("Regression Data")
plt.show()
```

```python
# Lets apply OLS
from sklearn.linear_model import LinearRegression
```

```python
from sklearn.linear_model import LinearRegression

reg = LinearRegression()
reg.fit(x.reshape(-1, 1), y)
```

```python
reg.intercept_
```

```python
plt.scatter(x, y)
plt.plot(x, reg.predict(x.reshape(-1, 1)), color='red')
```

```python
# lets apply Gradient Descent assuming slops is constant m = 78.35
# and lets's is assume the starting value for intercept b = 0
y_pred = ((78.35 * x) + 0).reshape(100)
```

```python
plt.scatter(x, y)
plt.plot(x, reg.predict(x.reshape(-1, 1)), color='red', label='OLS')
plt.plot(x, y_pred, color='#00a65a', label='b=0')
plt.legend()
plt.show()


```

```python
m = 78.35
b = 0

loss_slope = -2*np.sum(y-m*x.ravel()-b)
loss_slope
```

```python
# Lets take learning rate = 0.1
lr = 0.1

step_size  = loss_slope*lr
step_size
```

```python
# Calculating the new intercept
b = b - step_size
b
```

```python
y_pred1 = ((78.35 * x) + b).reshape(-1)

plt.scatter(x, y)
plt.plot(x, reg.predict(x.reshape(-1, 1)), color='black', label='OLS')
plt.plot(x, y_pred1, color='#00a65a', label='b = {}'.format(b))
plt.plot(x, y_pred, color='#A3E4D7', label='b = 0')
plt.legend()
plt.show()
```

```python
# Iteration 2
loss_slope = -2 * np.sum(y - m*x.ravel() -b)
loss_slope
```

```python
step_size = loss_slope*lr
step_size
```

```python
b = b - step_size
b
```

```python
y_pred2 = ((78.35 * x) + b).reshape(-1)

plt.scatter(x, y)
plt.plot(x, reg.predict(x.reshape(-1, 1)), color='red', label='OLS')
plt.plot(x, y_pred2, color='#00a65a', label='b = {}'.format(b))
plt.plot(x, y_pred1, color='blue', label='b = {}'.format(b))
plt.plot(x, y_pred, color='#A3E4D7', label='b = 0')
plt.legend()
plt.show()
```

```python
#  Iteration 3
loss_slope = -2 * np.sum(y-m*x.ravel() -b)
loss_slope
```

```python
step_size  = loss_slope*lr
step_size
```

```python
b = b - step_size
b
```

```python
plt.scatter(x, y)
plt.plot(x, reg.predict(x.reshape(-1, 1)), color='red', label='OLS')
plt.plot(x, y_pred2, color='#00a65a', label='b = {}'.format(b))
plt.plot(x, y_pred1, color='blue', label='b = {}'.format(b))
plt.plot(x, y_pred, color='#A3E4D7', label='b = 0')
plt.legend()
plt.show()
```
