# without column transformation 

```python
from sklearn.impute import SimpleImputer

# Create the imputer object
si = SimpleImputer()

# Fit on the training data and transform it
x_train_fever = si.fit_transform(x_train[['fever']])

# Use the same fitted imputer to transform the test data
x_test_fever = si.transform(x_test[['fever']])

# Check the shape
x_train_fever
```

```python
from sklearn.preprocessing import OrdinalEncoder

# Create the encoder
oe = OrdinalEncoder(categories=[['Mild', 'Strong']])

# Fit and transform the training data
x_train_cough = oe.fit_transform(x_train[['cough']])

# Transform the test data using the same encoder
x_test_cough = oe.transform(x_test[['cough']])

# View the encoded training data
x_train_cough.shape
```

```python
from sklearn.preprocessing import OneHotEncoder

ohe = OneHotEncoder(drop='first', sparse_output=False)

# Fit on training data
x_train_gender_city = ohe.fit_transform(x_train[['gender', 'city']])

# Transform test data
x_test_gender_city = ohe.transform(x_test[['gender', 'city']])

print(x_train_gender_city.shape) 
```

# with column column transformation

```python
from sklearn.compose import  ColumnTransformer
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OrdinalEncoder, OneHotEncoder

transformer = ColumnTransformer(
    transformers=[
        ('tnf1', SimpleImputer(), ['fever']),
        ('tnf2', OrdinalEncoder(categories=[['Mild', 'Strong']]), ['cough']),
        ('tnf3', OneHotEncoder(drop='first', sparse_output=False), ['gender', 'city'])
    ],
    remainder='passthrough'
) 
```

```python
import pandas as pd

x_train_transform = pd.DataFrame(
    transformer.fit_transform(x_train)
)

x_train_transform.head()
```

```python
import pandas as pd

x_train_transform = pd.DataFrame(
    transformer.fit_transform(x_train),
    columns=transformer.get_feature_names_out()
)

x_train_transform.head()
```

```python
x_test_transform = pd.DataFrame(
    transformer.transform(x_test),
    columns=transformer.get_feature_names_out()
)

x_test_transform.head()
```
